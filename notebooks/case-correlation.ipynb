# Wardlight local case analysis
This notebook reads only bundled synthetic JSON. It does not connect to Azure, MCP or Copilot. Run from this repository’s notebooks folder or repository root.
Compare counts and event IDs with data/answer-key.json. A file-access record does not prove a completed transfer.

```python
from pathlib import Path
from collections import Counter
import json
root = Path.cwd() if (Path.cwd()/"data/events.json").exists() else Path.cwd().parent
events = json.loads((root/"data/events.json").read_text(encoding="utf-8"))
counts = Counter(e["Account"] for e in events if e["Action"] == "SignInFailed")
print(dict(counts))
for event in sorted(events, key=lambda e: e["TimeGenerated"]):
    if counts[event["Account"]] >= 5 and event["Action"] in ("SignInSucceeded", "ProcessCreated", "FileAccessed"):
        print(event["TimeGenerated"], event["EventId"], event["Action"])
```
